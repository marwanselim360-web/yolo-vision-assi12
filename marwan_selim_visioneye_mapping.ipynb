```python
import cv2
import numpy as np
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

frame = np.zeros((480, 640, 3), dtype=np.uint8)
frame[:] = (45, 45, 45)
cv2.rectangle(frame, (150, 220), (240, 340), (0, 200, 255), -1)
cv2.rectangle(frame, (380, 180), (490, 320), (255, 180, 0), -1)

results = model(frame, verbose=False)[0]
map_canvas = np.zeros((480, 480, 3), dtype=np.uint8)
map_canvas[:] = (25, 25, 25)

for box in results.boxes.xyxy.cpu().numpy():
  x1, y1, x2, y2 = box[:4]
  cx = int((x1 + x2) / 2)
  cy = int(y2)
  mx = int((cx / 640) * 480)
  my = int((cy / 480) * 480)
  cv2.circle(map_canvas, (mx, my), 8, (0, 255, 0), -1)
  cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)

combined = np.hstack((frame, cv2.resize(map_canvas, (480, 480))))
cv2.imwrite("visioneye_map_output.png", combined)
```
