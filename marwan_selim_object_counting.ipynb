```python


fourcc = cv2.VideoWriter_fourcc(*"XVID")
writer = cv2.VideoWriter("sample_video.mp4", fourcc, 20.0, (640, 480))

for i in range(100):
  frame = cv2.UMat(480, 640, cv2.CV_8UC3).get()
  frame[:] = (30, 30, 30)
  x = 50 + i * 5
  cv2.rectangle(frame, (x, 200), (x + 80, 260), (0, 255, 0), -1)
  writer.write(frame)

writer.release()
```

```python
import cv2
from ultralytics.solutions import ObjectCounter

video_path = "sample_video.mp4"
cap = cv2.VideoCapture(video_path)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
line_points = [(100, int(height * 0.5)), (width - 100, int(height * 0.5))]

counter = ObjectCounter(
    show=False,
    region=line_points,
    model="yolov8n.pt",
)

fourcc = cv2.VideoWriter_fourcc(*"XVID")
out = cv2.VideoWriter(
    "counted_output.avi", fourcc, fps if fps > 0 else 20.0, (width, height)
)

while cap.isOpened():
  success, frame = cap.read()
  if not success:
    break

  results = counter(frame)
  out.write(results.plot_im)

cap.release()
out.release()
```
